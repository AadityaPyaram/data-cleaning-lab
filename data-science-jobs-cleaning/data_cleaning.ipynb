*Size*

```python
df['Size']
```

```python
df['Size'].apply(lambda x: x.split('to'))
```

```python
# we can't use apply function due to the presence of NaN values, so we'll use .str method

df['Size'].str.split('to').str[0].astype('Int32')
```

```python
# now we found out that some values or not in range but like '10000+'

df['min_employees'] = (
    df['Size']
    .str.extract(r'(\d+)')[0]
    .astype('Int64')
)
```

```python
df['max_employees'] = (
    df['Size']
    .str.extract(r'to\s*(\d+)')[0]
    .astype('Int64')
)
```

*Revenue*

```python
df['Revenue'].sample(10)
```

```python
# there's some values as x+ B, some as x to y M/B
# we're creating 2 new columns - min_revenue, max_revenue
# but the catch is the values units will be in 'million'
# we'll follow the 1B = 1000M
# while also keeping in mind that values differ in type

df['Revenue'].sample(10)
```

```python
# creating a function to get min_revenue

def min_revenue(x):

  # for NaN values
  if pd.isna(x):
    return np.nan

  # for $x+ M/B values
  if '+' in x:
    if 'M' in x:
      value = float(x.split('+')[0].replace('$', '').strip())
    else:
      value = float(x.split('+')[0].replace('$', '').strip())*1000
    return value

  # for $x M/B to $y M/B values
  if 'to' in x:
    value = x.split('to')[0].replace('$', '').strip()
    if 'M' in value:
      value = float(value.replace('M', '').strip())
    elif 'B' in value:
      value = float(value.replace('B', '').strip())*1000
    value = float(value)
    return value

df['min_revenue($M)'] = df['Revenue'].apply(min_revenue)
```

```python
# creating a function to get max_revenue

def max_revenue(x):

  # for NaN values
  if pd.isna(x):
    return np.nan

  # for $x+ M/B
  if '+' in x:
    return np.nan

  # for $x M/B to $y M/B values
  if 'to' in x:
    value = x.split('to')[1].replace('$', '').strip()
    if 'M' in value:
      value = float(value.replace('M', '').strip())
    elif 'B' in value:
      value = float(value.replace('B', '').strip())*1000
    value = float(value)
    return value

df['max_revenue($M)'] = df['Revenue'].apply(max_revenue)
```
